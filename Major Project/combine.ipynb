```python
import pandas as pd
import mysql.connector
```

```python
customers = pd.read_csv(r"Processed Datasets/customers.csv", encoding="latin-1")
products = pd.read_csv(r"Processed Datasets/products.csv", encoding="latin-1")
sales = pd.read_csv(r"Processed Datasets/sales.csv", encoding="latin-1")
```

```python
print("Customers:", len(customers))
print("Products:", len(products))
print("Sales:", len(sales))
```

```python
conn = mysql.connector.connect(
    host="localhost",
    port=3306,
    user="root",
    password="Fateh2005",
    database="eios_demo"
)
```

```python
cursor = conn.cursor()
```

```python
customer_data = list(
    customers[
        ["customer_id", "customer_name", "segment", "country", "state"]
    ].itertuples(index=False, name=None)
)

cursor.executemany("""
    INSERT INTO customers
    (customer_id, customer_name, segment, country, state)
    VALUES (%s, %s, %s, %s, %s)
""", customer_data)

conn.commit()

print("Customers inserted:", cursor.rowcount)
```

```python
product_data = list(
    products[
        [
            "product_key",
            "product_id",
            "product_name",
            "category",
            "sub_category"
        ]
    ].itertuples(index=False, name=None)
)

cursor.executemany("""
    INSERT INTO products
    (product_key, product_id, product_name, category, sub_category)
    VALUES (%s, %s, %s, %s, %s)
""", product_data)

conn.commit()

print("Products inserted:", cursor.rowcount)
```

```python
sales_data = list(
    sales[
        [
            "sales_record_id",
            "order_id",
            "customer_id",
            "product_key",
            "order_date",
            "ship_date",
            "ship_mode",
            "market",
            "region",
            "quantity",
            "sales",
            "discount",
            "profit",
            "shipping_cost",
            "order_priority",
            "year"
        ]
    ].itertuples(index=False, name=None)
)

cursor.executemany("""
    INSERT INTO sales (
        sales_record_id,
        order_id,
        customer_id,
        product_key,
        order_date,
        ship_date,
        ship_mode,
        market,
        region,
        quantity,
        sales,
        discount,
        profit,
        shipping_cost,
        order_priority,
        year
    )
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
""", sales_data)

conn.commit()

print("Sales inserted:", cursor.rowcount)
```

```python
cursor.execute("""
SELECT
    s.order_id,
    c.customer_name,
    p.product_name,
    s.sales,
    s.profit
FROM sales s
JOIN customers c
    ON s.customer_id = c.customer_id
JOIN products p
    ON s.product_key = p.product_key
LIMIT 10
""")

for row in cursor.fetchall():
    print(row)
```
